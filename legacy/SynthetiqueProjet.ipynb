# Fichier Synthétique du projet NLP de Star Wars

## 0. Objectif du projet 


Ce projet consiste à effectuer un traitement des données textuelles et à y appliquer des méthodes de fouille de textes.
Pour cela, j’ai choisi d’étudier le basculement d’Anakin Skywalker vers le côté obscur dans la saga Star Wars.

J’ai commencé par scraper les scripts en anglais des épisodes concernés, puis appliqué des algorithmes de prétraitement et d’analyse textuelle sur ces données. Une fois les textes nettoyés et formatés, j’ai analysé le lexique utilisé par les gentils et les méchants de la saga afin de le comparer à celui employé par notre sujet, Anakin.

Une fois cette analyse terminée, j’ai utilisé un modèle prédictif sur les données pour tenter de déterminer si les phrases prononcées par Anakin sont associées au côté lumineux ou au côté obscur de la Force.

Dans Star Wars, la distinction entre les gentils et les méchants repose en grande partie sur la gestion des émotions. Par conséquent, nous avons également étudié les sentiments exprimés par Anakin dans chacune de ses phrases pour mieux comprendre son évolution psychologique au fil des épisodes.


## 1. Importation des données

Pour réaliser ce projet, j’ai commencé par scraper les scripts en anglais des trois premiers épisodes de Star Wars. J’ai ensuite appliqué un prétraitement pour nettoyer et harmoniser les textes.

Pour standardiser les noms des personnages, j’ai utilisé une fonction de correction dynamique basée sur la similarité des chaînes. Toutefois, j’ai rencontré des difficultés avec la gestion des caractères spéciaux et l’extraction précise des phrases par personnage en raison du format irrégulier des scripts.

Malgré ces défis, les textes structurés ont permis d’analyser le lexique et les sentiments exprimés par Anakin.

Une fois structuré, j'ai regroupé les textes dans trois listes, l'une contenant tous els scripts des gentils, l'autres des méchants et la dernière avec celui de Anakin.

## 2. Analyse des Scripts

Pour l’analyse des scripts, j’ai commencé par extraire les dialogues des personnages afin de les structurer et les organiser. J’ai ensuite procédé à une analyse lexicale pour comparer le vocabulaire utilisé par les gentils et les méchants de la saga avec celui d’Anakin Skywalker.

Cette étape m’a permis d’observer les mots récurrents, les différences de ton et les évolutions lexicales qui accompagnent la transformation d’Anakin vers le côté obscur. J’ai également appliqué des techniques d’analyse des sentiments pour identifier les émotions dominantes dans ses dialogues, comme la colère, la peur ou l’hésitation.

L’extraction et l’interprétation des résultats ont révélé des tendances marquantes dans le langage d’Anakin, en lien avec son passage progressif vers le côté obscur.

## 3. Analyse de Anakin

Dans la troisième partie, j’ai appliqué un modèle prédictif afin de déterminer si les phrases d’Anakin sont associées au côté lumineux ou au côté obscur de la Force. Pour cela, j’ai utilisé les données textuelles préalablement nettoyées et structurées.

Le modèle a été entraîné en s’appuyant sur le lexique des gentils et des méchants pour établir des critères de classification. Ensuite, j’ai évalué ses performances pour vérifier sa capacité à prédire correctement l’appartenance des dialogues.

Cette analyse m’a permis de mettre en évidence la transition linguistique et émotionnelle d’Anakin, traduisant son basculement progressif vers le côté obscur au fil des épisodes.

## Conclusion

Ce projet a analysé l’évolution d’Anakin Skywalker vers le côté obscur en étudiant ses dialogues à travers des techniques de traitement du language. L’analyse lexicale, des sentiments et la classification ont révélé une progression émotionnelle et linguistique marquant son basculement vers le côté obscur.
